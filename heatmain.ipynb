```python
import numpy as np
from matplotlib import pyplot
```

```python
length = 10
k = 0.89
temp_left = 100
temp_right = 200
total_sim_time = 10
```

```python
dx = 0.1
x_vector = np.linspace(0,length,int(length/dx))
```

```python
x_vector
```

```python
dt = 0.0001
t_vector = np.linspace(0,total_sim_time,int(total_sim_time/dt))
t_vector
```

```python
u = np.zeros([len(t_vector),len(x_vector)])
u[:,0] = temp_left
u[:,-1] = temp_right
u
```

```python
pyplot.plot(x_vector,u[0])
pyplot.ylabel("temp")
pyplot.xlabel("x")
pyplot.show()
```

```python
for t in range(1,len(t_vector)-1):
    for x in range(1,len(x_vector)-1):
        u[t+1,x] = ((k * (dt/dx**2)) * (u[t,x+1] - 2* u[t,x] + u[t,x-1])) + u[t,x]
```

```python
pyplot.plot(x_vector,u[95000])
pyplot.ylabel("temp")
pyplot.xlabel("x")
pyplot.show()
```
